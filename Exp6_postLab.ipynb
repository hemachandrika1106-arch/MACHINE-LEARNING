# PCA and Clustering on Titanic Dataset

**Objective:** Perform PCA and K-Means clustering on the Titanic dataset.


```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
import warnings
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.decomposition import PCA
```

```python
from sklearn.cluster import KMeans
```

```python
warnings.filterwarnings("ignore")
```

```python
print("Libraries imported successfully")
```

```python
titanic = sns.load_dataset("titanic")
```

```python
print("Dataset loaded successfully")
```

```python
print("Dataset Shape:", titanic.shape)
```

```python
print("\nFirst 5 Rows:")
```

```python
display(titanic.head())
```

```python
data = titanic[["pclass", "age", "sibsp", "parch", "fare"]].copy()
```

```python
print("Selected Features:")
```

```python
display(data.head())
```

```python
print("\nDataset Shape:", data.shape)
```

```python
print("Missing Values Before Preprocessing:")
```

```python
print(data.isnull().sum())
```

```python
data["age"] = data["age"].fillna(data["age"].median())
```

```python
print("\nMissing Values After Preprocessing:")
```

```python
print(data.isnull().sum())
```

```python
scaler = StandardScaler()
```

```python
scaled_data = scaler.fit_transform(data)
```

```python
print("Data standardized successfully")
```

```python
print("\nFirst 5 Rows of Scaled Data:")
```

```python
print(scaled_data[:5])
```

```python
pca = PCA(n_components=2)
```

```python
pca_data = pca.fit_transform(scaled_data)
```

```python
print("PCA applied successfully")
```

```python
print("Shape after PCA:", pca_data.shape)
```

```python
print("\nExplained Variance Ratio:")
```

```python
print(pca.explained_variance_ratio_)
```

```python
pca_df = pd.DataFrame(pca_data, columns=["Principal Component 1", "Principal Component 2"])
print("First 5 Rows after PCA:")
display(pca_df.head())
```

```python
inertia = []
```

```python
for k in range(1, 8):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(pca_data)
    inertia.append(kmeans.inertia_)
```

```python
print("Inertia Values:")
```

```python
print(inertia)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(range(1, 8), inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("24EU02040 -> Elbow Method - Titanic Dataset")
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
clusters = kmeans.fit_predict(pca_data)
```

```python
pca_df["Cluster"] = clusters
```

```python
print("K-Means clustering completed successfully")
```

```python
print("\nNumber of Passengers in Each Cluster:")
```

```python
print(pca_df["Cluster"].value_counts().sort_index())
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="Principal Component 1", y="Principal Component 2",
                hue="Cluster", palette="Set1", alpha=0.7)
plt.title("24EU02040 -> Titanic Passenger Clusters using PCA and K-Means")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

```python
print("Cluster visualization generated successfully")
```

```python
print("Cluster Centers in PCA Space:")
```

```python
centers = pd.DataFrame(kmeans.cluster_centers_,
                       columns=["Principal Component 1", "Principal Component 2"])
```
